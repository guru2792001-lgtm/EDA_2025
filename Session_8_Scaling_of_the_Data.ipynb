```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df1=pd.read_csv(r"F:\Omkar_sir_files\CSV_Files\Visadataset (1).csv")
cat=df1.select_dtypes(include='object').columns
num=df1.select_dtypes(exclude='object').columns
num
```

```python
x=df1['prevailing_wage']
u=x.mean()
sd=x.std()
nr=x-u
df1['P_wage_z']=nr/sd
```

```python
df1[['prevailing_wage','P_wage_z']]
```

```python
plt.figure(figsize=(8,3))
plt.subplot(1,2,1).hist(df1['prevailing_wage'],bins=40)
plt.subplot(1,2,2).hist(df1['P_wage_z'],bins=40)
plt.show()

```

```python
x=df1['prevailing_wage']
minn=x.min()
maxx=x.max()
Nr=x-minn
Dr=maxx-minn
# wage_norm=Nr/Dr
df1['wage_norm']=Nr/Dr
```
