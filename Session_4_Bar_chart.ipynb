```python
import os
folder='Frequency_Data_2'
try:
    os.makedirs(folder)
except Exception as e:
    print(e)
try:
    cwd=os.getcwd()
    nwd=os.path.join(cwd,folder)
except Exception as e:
    print(e)

try:
    for i in catagorical:
        data=df1[i].value_counts()
        d1=data.values
        d2=data.keys()
        guru=pd.DataFrame(zip(d2,d1),columns=['column','count'])
        file_name=f'{i}_guru.csv'
        path=os.path.join(nwd,file_name)
        guru.to_csv(path,index=False)
        
except Exception as e:
    print(e)
```

## **Bar Charts**

- bar charts under the matplotlib
- plt.bar
- bar chart needs class names and class frequency
- class names we can get from vlue counts
- class frequency we can get from value count

```python
# plt.bar(d2,d1)
# plt.show()
```

```python
data=df1['continent'].value_counts()
data
```

```python
d1=data.values
d1
```

```python
d2=data.keys()
d2
```

```python
colors=['Red','Green','Yellow','Orange','Blue','Cyan']
plt.figure(figsize=(10,5))
plt.bar(d2,d1,color=colors)
plt.title("continent_BarChart")
plt.xlabel("Class Names")
plt.ylabel("Class Frequency")
plt.savefig("continent_BarChart.jpg")
plt.show() #show means closing only pic will be show
```

```python
for i in catagorical[1:]:
    data=df1[i].value_counts()
    d1=data.values
    d2=data.keys()
    # colors=['Red','Green','Yellow','Orange','Blue','Cyan']
    plt.figure(figsize=(10,5))
    plt.bar(d2,d1,color=colors)
    plt.title(f"{i}_BarChart")
    plt.xlabel("Class Names")
    plt.ylabel("Class Frequency")
    # plt.savefig(f"{i}_BarChart.jpg")
    plt.show() #show means closing only pic will be show
        


    
```
