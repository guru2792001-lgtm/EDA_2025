## **Step-1 : import pckages**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## **Step-2 : CReate a Dataframe using List**

```python
names=['Ramesh','Suresh','Mahesh']
pd.DataFrame(names)
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
pd.DataFrame(names,age)
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
pd.DataFrame(zip(names,age))
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
city=['kolhapur','hyd','pune']
pd.DataFrame(zip(names,age,city))
```

## **Step-3: Insert the columns**
- column value expect an idex or array-like

```python
names=['Ramesh','Suresh','Mahesh']
pd.DataFrame(names,columns=['Names'])
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
city=['kolhapur','hyd','pune']
cols=['Names','Age','city']
pd.DataFrame(zip(names,age,city),columns=cols)
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
city=['kolhapur','hyd','pune']
i=['a','b','c']
pd.DataFrame(zip(names,age,city),index=i)
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
city=['kolhapur','hyd','pune']
cols=['Names','Age','city']
i=['a','b','c']
pd.DataFrame(zip(names,age,city),index=i,columns=cols)
```

```python
names=['Guru','akash','tushar']
age=[23,34,45]
city=['kolhapur','buldhana','fulewadi']
cols=['Names','Age','city']
i=['a','b','c']
pd.DataFrame(zip(names,age,city),index=i,columns=cols)
```

## Step-5: Create a Dataframe using dictionary

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
city=['kolhapur','hyd','pune']

d={'names':['Ramesh','Suresh','Mahesh'],
'age':[20,21,22],
'city':['kolhapur','hyd','pune']}

pd.DataFrame(d)


#Keys behaves as columns names
# NO need of explacity mention
```

## Step-6 : Create a DataFrame using scalar values means with out list

```python
d={'names':'Ramesh',
'age':20,
'city':'kolhapur'}

pd.DataFrame(d)
```

```python
d={'names':'Ramesh',
'age':20,
'city':'kolhapur'}

pd.DataFrame(d,index=[1])
```

## **Step-10:How to update the column values**

## **Step-11:How to Drop the specific columns**

- by default axis=0 represents index
- if you provide the column change axis=1
- inplace=True then Data Frame will be overwrite

```python
d1.drop('Age',axis=1,
       inplace=True)
```

```python
d1
```

```python
names=['Ramesh','Suresh','Mahesh']
age=[20,21,22]
city=['kolhapur','hyd','pune']
cols=['Names','Age','city']
i=['a','b','c']
pd.DataFrame(zip(names,age,city),index=i,columns=cols)
```
