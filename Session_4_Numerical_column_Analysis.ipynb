## **Emperical Rule**

```python
wage_mean=wage_data.mean()
sigma=wage_data.std()
lb=wage_mean-1*sigma
ub=wage_mean+1*sigma
con1=wage_data>lb
con2=wage_data<ub
con=con1 & con2
(len(df1[con])==68*len(df1)/100)
```

```python
def Emperical_rule(k,n):
    wage_mean=wage_data.mean()
    sigma=wage_data.std()
    lb=wage_mean-k*sigma
    ub=wage_mean+k*sigma
    con1=wage_data>lb
    con2=wage_data<ub
    con=con1 & con2
    print(len(df1[con])==n*len(df1)/100)
Emperical_rule(3,99.7)
```
